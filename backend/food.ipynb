```python
training_foods = ['Amaranth', 'Apple', 'Banana', 'Beetroot', 
                   'Bell pepper', 'Bitter Gourd', 'Blueberry', 
                   'Bottle Gourd', 'Broccoli', 'Cabbage', 'Cantaloupe', 
                   'Capsicum', 'Carrot', 'Cauliflower', 'pepper', 
                   'Coconut', 'Corn', 'Cucumber', 'Dragon fruit', 'Eggplant', 
                   'Fig', 'Garlic', 'Ginger', 'Grapes', 'Jalepeno', 'Kiwi', 
                   'Lemon', 'Mango', 'Okra', 'Onion', 'Orange', 'Paprika', 
                   'Pear', 'Peas', 'Pineapple', 'Pomegranate', 'Potato', 
                   'Pumpkin', 'Raddish', 'Raspberry', 'Gourd', 'Soy beans', 
                   'Spinach', 'Strawberry', 
                   'Sweetcorn', 'Sweetpotato', 'Tomato', 'Turnip', 'Watermelon'
]

training_foods_lower = [f.lower() for f in training_foods]

df = pd.read_csv("recipes_ingredients.csv")

def get_matching_ingredients(ner_value):
    """Return list of matching ingredients for a row."""
    if pd.isna(ner_value):
        return []
    ner_lower = ner_value.lower()
    return [food for food in training_foods_lower if food in ner_lower]

# Track how many recipes we've collected per ingredient
ingredient_counts = {food: 0 for food in training_foods_lower}
selected_indices = set()

for idx, row in df.iterrows():
    matches = get_matching_ingredients(row['ingredients'])
    for food in matches:
        if ingredient_counts[food] < 5:
            selected_indices.add(idx)
            ingredient_counts[food] += 1

filtered_df = df.loc[sorted(selected_indices)]

print(f"Original rows: {len(df)}")
print(f"Filtered rows: {len(filtered_df)}")
print("\nRecipes per ingredient:")
for food, count in ingredient_counts.items():
    if count > 0:
        print(f"  {food}: {count}")

filtered_df.to_csv('filtered_recipes.csv', index=False)
print("\nSaved to filtered_recipes.csv")
```
