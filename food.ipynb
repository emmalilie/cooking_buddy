```python
import pandas as pd
```

```python
training_foods = [
    'Apple',
    'Banana',
    'Bell pepper',
    'Broccoli',
    'Cabbage',
    'Carrot',
    'Cauliflower',
    'Chilli pepper',
    'Corn',
    'Cucumber',
    'Eggplant',
    'Garlic',
    'Ginger',
    'Grapes',
    'Lemon',
    'Mango',
    'Onion',
    'Orange',
    'Pear',
    'Pineapple',
    'Potato',
    'Pumpkin',
    'Spinach',
    'Strawberry',
    'Sweetcorn',
    'Sweetpotato',
    'Tomato'
]
```

```python
df = pd.read_csv('nutrition.csv')
```

```python
nutrition_info = []

for food in training_foods:
    search_term = food.lower()
    matches = df[df['name'].str.lower().str.contains(search_term, na=False)]
    
    if not matches.empty:
        raw_match = matches[matches['name'].str.contains('raw', case=False, na=False)]
        match = raw_match.iloc[0] if not raw_match.empty else matches.iloc[0]
        
        nutrition_info.append({
            'food': food,
            'name': match['name'],
            'calories': match['calories'],
            'protein': match['protein'],
            'carbohydrate': match['carbohydrate'],
            'fiber': match['fiber']
        })

result = pd.DataFrame(nutrition_info)
result
```

```python
result.to_csv('food_nutrition.csv', index=False)
```
